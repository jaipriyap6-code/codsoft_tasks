Find total Sales and Profit

```python
print("Total Sales:", ds['Sales'].sum())
print("Total Profit:", ds['Profit'].sum())
```

Find average Sales and Profit

```python
print("Average Sales:", ds['Sales'].mean())
print("Average Profit:", ds['Profit'].mean())
```

Find highest and lowest Sales

```python
print("Highest Sales:", ds['Sales'].max())
print("Lowest Sales:", ds['Sales'].min())
```

Which category generates the highest sales

```python
ds.groupby('Category')['Sales'].sum()
```

```python
ds.groupby('Category')['Sales'].sum().sort_values(ascending=False)
```

Which category generates the highest profit?

```python
ds.groupby('Category')['Profit'].sum().sort_values(ascending=False)
```

Which customer segment generates the most sales?

```python
ds.groupby('Segment')['Sales'].sum().sort_values(ascending=False)
```

Profit by Customer Segment

```python
ds.groupby('Segment')['Profit'].sum().sort_values(ascending=False)
```

which city generates the most sales.

```python
ds.groupby('City')['Sales'].sum().sort_values(ascending=False)
```

Sales by State

```python
ds.groupby('State')['Sales'].sum().sort_values(ascending=False)
```

Convert Order Date

```python
ds['Order_Date'] = pd.to_datetime(ds['Order_Date'])
```

```python
ds['Order_Date'].dtype
```

Analyze sales over time

```python
ds['Year'] = ds['Order_Date'].dt.year
```

```python
ds.groupby('Year')['Sales'].sum()
```

Check relationship between Sales and Profit

```python
ds[['Sales', 'Profit']].corr()
```

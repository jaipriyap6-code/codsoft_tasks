```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/superstore_sales_eda.csv')
```

Convert Order Date

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

```python
df['Year'] = df['Order_Date'].dt.year
```

BAR CHART

```python
sales_category = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8,5))
plt.bar(sales_category.index, sales_category.values)

plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')

plt.show()
```

LINE CHART

```python
sales_year = df.groupby('Year')['Sales'].sum()

plt.figure(figsize=(8,5))
plt.plot(sales_year.index, sales_year.values, marker='o')

plt.title('Sales Trend by Year')
plt.xlabel('Year')
plt.ylabel('Total Sales')

plt.show()
```

PIE CHART

```python
segment_sales = df.groupby('Segment')['Sales'].sum()

plt.figure(figsize=(7,7))
plt.pie(
    segment_sales.values,
    labels=segment_sales.index,
    autopct='%1.1f%%'
)

plt.title('Sales Distribution by Customer Segment')

plt.show()
```

HISTOGRAM

```python
plt.figure(figsize=(8,5))

plt.hist(df['Sales'], bins=10)

plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')

plt.show()
```

SCATTER PLOT

```python
plt.figure(figsize=(8,5))

plt.scatter(df['Sales'], df['Profit'])

plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')

plt.show()
```

Add a Seaborn visualization

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Category', y='Profit', estimator=sum)
plt.bar(
    sales_category.index,
    sales_category.values,
    color=['skyblue', 'orange', 'green'])
plt.title('Total Profit by Category')
plt.xlabel('Category')
plt.ylabel('Total Sale')
plt.show()
```

add color

```python
sales_category = df.groupby('Category')['Sales'].sum()
plt.figure(figsize=(8,5))
bars=plt.bar(
    sales_category.index,
    sales_category.values,
    color=['skyblue', 'orange', 'green'])
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{bar.get_height():,.0f}',
        ha='center',
        va='bottom')
plt.show()
```

Sales increased from 2023 to 2024, indicating positive sales growth during the period.

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Year'] = df['Order_Date'].dt.year
sales_year = df.groupby('Year')['Sales'].sum()
plt.figure(figsize=(8,5))
plt.plot(sales_year.index, sales_year.values, marker='o')
plt.title('Sales Trend by Year')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.show()
#2023 → 124,768.09
#2024 → 153,468.95
```

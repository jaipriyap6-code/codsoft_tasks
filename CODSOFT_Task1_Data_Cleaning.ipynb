# **Check the median values first**

```python
"Median Age:", emp['Age'].median()
```

 change the data type of sal beacuse currently data type of sal is object

```python
emp['Salary'] = pd.to_numeric(emp['Salary'], errors='coerce')
```

```python
emp['Salary'].median()
```

```python
emp['Salary'] = emp['Salary'].fillna(59500)
```

```python
emp['Salary'].isnull().sum()
```

# **find median age**

```python
emp['Age'].median()
```

# **use that number**

```python
emp['Age'] = emp['Age'].fillna(30)
```

check

```python
emp['Age'].isnull().sum()
```

fix missing in city col

```python
emp['City'] = emp['City'].fillna('Unknown')
```

```python
emp['City'].isnull().sum()
```

Check all missing values

```python
emp.isnull().sum()
```

check for duplicates

```python
emp.duplicated().sum()
```

Check Department values

```python
emp['Department'].unique()
```

Fix Department

```python
emp['Department'] = emp['Department'].str.strip()
```

```python
emp['Department'] = emp['Department'].replace({
    'it': 'IT',
    'sales': 'Sales',
    'Human Resources': 'HR'
})
```

```python
emp['Department'].unique()
```

fix gender

```python
emp['Gender'].unique()
```

```python
emp['Gender'] = emp['Gender'].replace({
    'female': 'Female'
})
```

```python
emp['Gender'].unique()
```

```python
emp.dtypes
```
